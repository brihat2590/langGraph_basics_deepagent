# **First Graph**


```python
from typing import TypedDict, List

# Define the schema for a graph node with a name and a message
class Graph_schema(TypedDict):
    name : str
    message: str



# define the nodes as a functiion then we later attach

def Welcome(Graph_schema: Graph_schema) -> Graph_schema:
    return {
        "name": "Welcome Node",
        "message": "This is the welcome message."
    }

```

```python
from langgraph.graph import StateGraph, END, START
from IPython.display import display, Image

graph=StateGraph(Graph_schema)

#adding nodes
graph.add_node("Welcome",Welcome)

#adding edges
graph.add_edge(START,"Welcome")
graph.add_edge("Welcome",END)


#compiling the graph
first_graph=graph.compile()
Image(first_graph.get_graph().draw_mermaid_png())
```

```python
# invoke the compiled graph, which runs the Welcome node
result = first_graph.invoke({"name": "", "message": ""})
print(result)
```
