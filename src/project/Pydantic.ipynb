## **Pydantic Schema**

```python
from pydantic import BaseModel, Field
from typing import List

from langchain_groq import ChatGroq
from dotenv import load_dotenv
import os

load_dotenv()

api_key = os.getenv("GROQ_API_KEY")

llm = ChatGroq(
    model="openai/gpt-oss-120b",
    api_key=api_key,
    temperature=0
)


class LearningPlan(BaseModel):
    concepts: List[str] = Field(
        description="Important concepts that should be explained"
    )

    example: str = Field(
        description="A practical technical example"
    )

    common_mistakes: List[str] = Field(
        description="Common mistakes beginners make"
    )


# Configure structured output
structured_llm = llm.with_structured_output(LearningPlan)

# Invoke
response = structured_llm.invoke(
    "Create a learning plan for Model Context Protocol."
)

print(response)
```

```python
class graph_schema(BaseModel):
    title: str = Field("Title", description="The title of the graph")
    description: str = Field("Description", description="The description of the graph")
    body: str = Field("Body", description="The body of the graph")


demo_object=graph_schema(
    title="Sample Graph",
    description="This is a sample graph description.",
    body="This is the body of the sample graph."

)
print(demo_object)
```

```python
from langchain_core.prompts import ChatPromptTemplate
prompt = ChatPromptTemplate.from_messages([
    (
        "system",
        "You are a precise programming teacher."
    ),
    (
        "human",
        """
        Explain {topic} for a {level} developer.

        Use:
        - A simple definition
        - A short code example
        - One common mistake
        """
    )
])

chain = prompt | llm
chain.invoke({"topic": "Artificial Intelligence", "level": "beginner"})
```

```python
from langchain_community.tools import DuckDuckGoSearchRun
from langchain.tools import tool

@tool(description="Search the web using DuckDuckGo")
def duck_duck_go_search(query: str)->str:
    duck_search = DuckDuckGoSearchRun()
    return duck_search.invoke(query)

    
```

```python
tools=[duck_duck_go_search]
llm_with_tools = llm.bind_tools(tools)
response = llm_with_tools.invoke("what is the latest news on AI? use the duck_duck_go_search tool")
print(response.tool_calls)
print(response.content)
```
